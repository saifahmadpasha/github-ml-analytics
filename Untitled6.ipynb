```python
import pandas as pd
import requests

# Step 1: Collect API Data
url = "https://api.github.com/search/repositories?q=machine+learning&sort=stars&order=desc&per_page=100"
response = requests.get(url)
data = response.json()

# Step 2: Create the Dataset
df = pd.DataFrame(data['items'])
print("Initial Dataset Shape:", df.shape)
print(df.info())

# Step 3: Select Required Columns
required_cols = [
    'name', 'owner', 'language', 'stargazers_count',
    'forks_count', 'watchers_count', 'open_issues_count',
    'created_at', 'updated_at', 'license'
]
df = df[required_cols]

# Step 4: Extract Required Values from Nested Columns
df['owner'] = df['owner'].apply(lambda x: x['login'] if isinstance(x, dict) and 'login' in x else None)
df['license'] = df['license'].apply(lambda x: x['name'] if isinstance(x, dict) and 'name' in x else 'No License')

# Step 5: Handle Missing Values
df['language'] = df['language'].fillna('Unknown')
df['owner'] = df['owner'].fillna('Unknown')

# Step 6: Handle Duplicate Records
df = df.drop_duplicates()

# Step 7: Prepare the Dataset (Convert dates and rename columns)
df['created_at'] = pd.to_datetime(df['created_at']).dt.date
df['updated_at'] = pd.to_datetime(df['updated_at']).dt.date

df = df.rename(columns={
    'stargazers_count': 'stars',
    'forks_count': 'forks',
    'watchers_count': 'watchers',
    'open_issues_count': 'open_issues',
    'created_at': 'created_date',
    'updated_at': 'updated_date'
})

# Step 8: Save and Verify the Dataset
file_name = "github_projects.csv"
df.to_csv(file_name, index=False)

# Verification step
verified_df = pd.read_csv(file_name)
print("\n--- Dataset Verification ---")
print("Saved file shape:", verified_df.shape)
print("Columns in saved CSV:", verified_df.columns.tolist())
print("\nFirst 5 Rows:")
print(verified_df.head())
```

```python
import pandas as pd
import sqlite3
import matplotlib.pyplot as plt

# Step 1: Create the SQLite Database & Import Data
conn = sqlite3.connect("github_data.db")
df = pd.read_csv("github_projects.csv")
df.to_sql("Repositories", conn, if_exists="replace", index=False)

print("--- Step 2: Filtering & Searching Queries ---")
# Repositories with more than 10,000 stars
q_stars = "SELECT name, stars, language FROM Repositories WHERE stars > 10000;"
print("\nRepos with > 10k stars:\n", pd.read_sql_query(q_stars, conn).head())

# Repositories containing the word 'Machine'
q_search = "SELECT name, stars FROM Repositories WHERE name LIKE '%Machine%';"
print("\nRepos with 'Machine' in name:\n", pd.read_sql_query(q_search, conn).head())


print("\n--- Step 3: Logical Operators ---")
# Query 1 using AND, NOT
q_logical1 = """
SELECT name, stars, language
FROM Repositories
WHERE stars > 5000 AND NOT language = 'Unknown';
"""
print("\nQuery 1 (AND, NOT):\n", pd.read_sql_query(q_logical1, conn).head())

# Query 2 using OR, AND
q_logical2 = """
SELECT name, stars, language
FROM Repositories
WHERE (language = 'Python' OR language = 'C++') AND stars > 20000;
"""
print("\nQuery 2 (OR, AND):\n", pd.read_sql_query(q_logical2, conn).head())


print("\n--- Step 4: Sorting & Limiting Queries ---")
# Top 10 most starred repositories
q_top10 = "SELECT name, stars, language FROM Repositories ORDER BY stars DESC LIMIT 10;"
print("\nTop 10 Starred Repositories:\n", pd.read_sql_query(q_top10, conn))


print("\n--- Step 5: Aggregate Functions ---")
# Total repos and average stars
q_agg = "SELECT COUNT(*) as total_repositories, AVG(stars) as average_stars FROM Repositories;"
print("\nAggregate Summary:\n", pd.read_sql_query(q_agg, conn))


print("\n--- Step 6: Grouping Analysis ---")
# Group by language with > 5 repositories
q_group = """
SELECT language, COUNT(*) as repo_count
FROM Repositories
GROUP BY language
HAVING repo_count > 5
ORDER BY repo_count DESC;
"""
print("\nLanguages with > 5 Repositories:\n", pd.read_sql_query(q_group, conn))


# Step 7: Create Python Visualizations

# Chart 1: Top 10 Most Popular Repositories
top10_df = pd.read_sql_query("SELECT name, stars FROM Repositories ORDER BY stars DESC LIMIT 10;", conn)
plt.figure(figsize=(10, 5))
plt.barh(top10_df['name'], top10_df['stars'], color='teal')
plt.gca().invert_yaxis()
plt.title('Top 10 Most Popular Machine Learning Repositories')
plt.xlabel('Stars')
plt.ylabel('Repository Name')
plt.tight_layout()
plt.savefig('top_10_repositories.png')
plt.show()

# Chart 2: Repository Creation Trends Over Time
trend_df = pd.read_sql_query("SELECT strftime('%Y', created_date) as year, COUNT(*) as repo_count FROM Repositories GROUP BY year ORDER BY year;", conn)
plt.figure(figsize=(9, 4))
plt.plot(trend_df['year'], trend_df['repo_count'], marker='o', color='coral', linewidth=2)
plt.title('Repository Creation Trend Over Time')
plt.xlabel('Year')
plt.ylabel('Number of Repositories Created')
plt.grid(True, linestyle='--', alpha=0.6)
plt.tight_layout()
plt.savefig('creation_trends.png')
plt.show()

conn.close()
```
